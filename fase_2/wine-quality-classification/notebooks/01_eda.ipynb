```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('..\\data\\raw\\WineQT.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['quality'].value_counts().sort_index()
```

```python
df['quality_label'] = (df['quality'] >= 7).astype(int)

df['quality_label'].value_counts()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')

plt.show()
```

```python
sns.boxplot(x='quality_label', y='alcohol', data=df)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='quality_label',
    y='volatile acidity',
    data=df
)

plt.title('Volatile Acidity por Classe de Qualidade')
plt.xlabel('Qualidade do Vinho (0 = Baixa/Média | 1 = Alta)')
plt.ylabel('Volatile Acidity')

plt.show()
```

```python
sns.boxplot(
    x='quality_label',
    y='sulphates',
    data=df
)
```
